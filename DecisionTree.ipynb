```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import tree
```

```python
df = pd.read_csv('salaries.csv')
df.head(8)
```

```python
plt.scatter(df.company,df.salary_more_then_100k)
```

```python
inputs = df.drop('salary_more_then_100k',axis='columns')
target = df['salary_more_then_100k']
```

```python
inputs
```

```python
target
```

```python
from sklearn.preprocessing  import LabelEncoder
le_company = LabelEncoder()
le_job = LabelEncoder()
le_degree = LabelEncoder()
```

```python
inputs['le_company'] = le_company.fit_transform(inputs['company'])
inputs['le_job'] = le_job.fit_transform(inputs['job'])
inputs['le_degree'] = le_degree.fit_transform(inputs['degree'])
```

```python
inputs
```

```python
input_n = inputs.drop(['company','job','degree'],axis = 'columns')
```

```python
input_n
```

```python
model = tree.DecisionTreeClassifier()
model.fit(input_n,target)
```

```python
model.score(input_n,target)
```

```python
model.predict(input_n)
```
