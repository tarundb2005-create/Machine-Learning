```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.datasets import load_digits   #bultin dataset
```

```python
digits = load_digits()
```

```python
digits.data[0]
```

```python
plt.gray()     #accurately display single-channel 2D data or images as shades of black, white, and gray
```

```python
for i in range(4):
  plt.matshow(digits.images[i])
```

```python
digits.target[0:4]  #checking the sample output is match to the pic
```

```python
X_train,X_test,y_train,y_test = train_test_split(digits.data,digits.target,test_size=0.2)
```

```python
model = LogisticRegression()       #using logistic model
model.fit(X_train,y_train)
```

```python
X_test
```

```python
model.predict(X_test)  #predicting the test sample
```

```python
model.score(X_test,y_test)         #accuracy
```

```python
y_predicted = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix          #using confusion matrix for the inaccuracy data
cm = confusion_matrix(y_test,y_predicted)
cm
```

```python
import seaborn as sn                #used seaborn to plot the confusion matrix
plt.figure(figsize = (10,7))
sn.heatmap(cm, annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```
